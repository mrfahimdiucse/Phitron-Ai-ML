```python
N = int(input())
num = input().split()
for i in range(N):
    val = int(num[i])
    if val % 2 == 0:
        continue
    digit = []
    temp = val
    while temp > 0:
        digit.append(temp % 10)
        temp = temp // 10
        sum = 0
    for j in range(len(digit)):
        sum += digit[j]
    if sum>20:
        continue
    print(num[i],sum)

       
```
