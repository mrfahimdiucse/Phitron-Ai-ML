```python
n=int(input())
inp=input()
nums=inp.split()
even_sum=0
odd_sum=0
positive_sum=0
negative_sum=0
for i in range(n):
    X=int(nums[i])
    if X>0:
        positive_sum+=1
    elif X<0:
        negative_sum+=1

    if X%2==0:
        even_sum+=1
    else:
        odd_sum+=1
print("Even: ",even_sum)
print("Odd: ",odd_sum)
print("Positive: ",positive_sum)
print("Negative: ",negative_sum)
```

```python
T=int(input())
for i in range(T):
    N=int(input())
    if N==0:
        print(0)
        continue
    while N>0:
        print(N%10,end=" ")
        N=N//10
    print(end="\n")
    
```
