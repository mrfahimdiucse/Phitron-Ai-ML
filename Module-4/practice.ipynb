```python
# problem 3
mark=int(input("Enter Your Marks Here: "))
if mark>100 or mark<0:
    print("Invlid Input!")
elif mark>=80:
    print("A+")
elif mark>=70:
    print("A")
elif mark>=40:
    print("Pass")
else:
    print("Fail")
```

```python
# problem 7
sum=0
n=input().split()
for i in n:
    if int(i)<0:
        break
    else:
        sum+=int(i)
print("Total : ",sum)
```
