### list as stack and queue


```python
numbers = [10, 20, 30, 40, 9, 2, 4, 23, 25]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)
```

```python
# stack as list
list=[]
#stack er push kora
list.append(1)
list.append(2)
list.append(3)
list.append(4)
list.append(5)
print(list)
# stack er pop kora
print(list.pop())
print(list)
```

### Queue As list

```python
queue=[]
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)
queue.append(6)
print(queue)
print(queue.pop(0))
print(queue)
# print top element
print(queue[-1])
```
