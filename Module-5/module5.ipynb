```python
tup=(10,20,12,30,40)
floatTup=(10.2,15.2,40.22,22.255)
mixed=(10,20,2.00,"Str",55.50)
print(tup)
print(floatTup)
print(mixed)
```

```python
tup=(10,20,30,40,50,30,20,30)
lis=list(tup)
print(type(lis))
```

```python
mixed=(10,20,2.00,"Str",55.50)
#indexing
print(mixed[3])
#slicing
new=mixed[0:3]
print(new)
```

### Set--> unique element store kore, mutable,unordered

```python
A={1,2,3}
print(type(A),A)
```

```python
#empty set declare
B=set()
print(type(B))
```

```python
s={1,2,3,4,1,2,5}
print(s)
s.add(10)
print(s)
if 9 in s:
    print("3 is present")
else:
    print("Not Present")

for i in s:
    print(i)
s.pop()#1st er dik theke remove hoye jay 

print(s)
s.remove(10)
print(s)
```

```python
#set methods
A={1,2,3}
B={2,3,4,5}
print(A.union(B))
print(A.intersection(B))
```

```python
a={1,2,3}
b={1,2,3,4,5}
print(a.isdisjoint(b))
print(a.issubset(b))
```

### Dictionaries 

```python
dic={}
print(type(dic))
```

```python
dic={"name":"Sadik Hasan", "age": 23, "Address": "Dhaka"}
print(dic)
#access
print(dic["name"])
print(dic.get("age"))
```

```python
dic={"name":"Sadik Hasan", "age": 23, "Address": "Dhaka","age":30 ,"numbers":[10,20,30]}
print(dic)
dic["name"]="Sadik Hasan Fahim"
print(dic)
print(dic.get("numbers"))
```

```python
#methods
dic={"name":"Sadik Hasan", "age": 23, "Address": "Dhaka","age":30 ,"numbers":[10,20,30]}
#adding a value
print(dic)
dic["math_marks"]=80
print(dic)
dic.update({"Eng_marks":40})
print(dic)

#delete
del dic["math_marks"]
print(dic)

#copy
dic2=dic.copy()
print(dic2)
```

```python
#view methods
print(dic)
keys=dic.keys()#real time update hoy 
print(keys)
value=dic.values()#real time update hoy 
print(value)

items=dic.items()
print(items)
```

```python
#iteration
for key,value in dic.items():
    print(key,value)
```
