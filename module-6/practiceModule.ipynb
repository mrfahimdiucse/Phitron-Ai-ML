### String Problems
S1. Palindrome Check
Space এবং uppercase/lowercase উপেক্ষা করে palindrome যাচাই করো।

```python
s = "A man a plan a canal Panama"
print(s.translate(str.maketrans("", "")))
ss="".join(ch.lower() for ch in s if ch.isalnum())
print(ss)
print(ss[::-1])
print(ss==ss[::-1])
```

```python
ss="".join(ch for ch in s if ch.isalnum())
print(ss)
```

```python
#problem 2
def batch_generator(batch_size,data):
    if batch_size < 1:
        return 
    for i in range(0, len(data), batch_size):
        yield data[i : i + batch_size]
data=list(map(int,input().split()))
batch_size=int(input())
for batch in batch_generator(batch_size,data):
    print(batch)
```
