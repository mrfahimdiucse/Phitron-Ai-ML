```python
def check_pass(marks):
    """
Checks if the student's marks meet the passing criteria (>= 40)
    """
    if marks>=40:
        return True
    else:
        return False

marks = int(input())
mark=check_pass(marks)
if mark:
    print("Pass")
else:
    print("Fail")
```

```python
def model_info(name, layers, lr=0.01):
    """Returns the formatted model registration info string."""
    return f"Model: {name}, Layers: {layers}, LR: {lr}"
t = int(input())
for _ in range(t):
    parts = input().split()
    req_type = parts[0]

    if req_type == "1":
        name, layers = parts[1], parts[2]
        print(model_info(name, layers))

    elif req_type == "2":
        name, layers = parts[1], parts[2]
        print(model_info(layers=layers, name=name))

    elif req_type == "3":
        name, layers, lr = parts[1], parts[2], parts[3]
        print(model_info(name, layers, lr))
```

```python
def batch_generator(data, batch_size):
    """Generates batches of a specified size from the data list.

    Yields nothing if batch_size < 1.
    """
    if batch_size < 1:
        return  

    for i in range(0, len(data), batch_size):
        yield data[i : i + batch_size]


def batch_mean(batch):
    """Calculates and returns the mean (average) of the numbers in a batch."""
    return sum(batch) / len(batch)

n, batch_size = map(int, input().split())
data = list(map(int, input().split()))

for batch in batch_generator(data, batch_size):
    avg = batch_mean(batch)
    print(f"{batch} mean={avg}")
```
